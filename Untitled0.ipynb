```python
dna = 'ATGCGATACGCTTACGATCGATCGCTAGCTAGCTAGCTGATCGA'
count_A = dna.count('A');
count_T = dna.count('T');
count_G = dna.count('G');
count_C = dna.count('C')
print(f'A: {count_A}');
print(f'T: {count_T}');
print(f'G: {count_G}');
print(f'C: {count_C}');
gc = (count_G + count_C) / len(dna) * 100;
print(f'GC Content: {gc:.2f}%')
```

```python
import matplotlib.pyplot as plt

# DNA sequence
dna = "ATGCGATACGCTTACGATCGATCGCTAGCTAGCTAGCTGATCGA"

# Count each base
bases = ["A", "T", "C", "G"]
counts = [dna.count(base) for base in bases]

# Plot bar chart
plt.bar(bases, counts)

plt.xlabel("DNA Bases")
plt.ylabel("Count")
plt.title("DNA Base Counts")

plt.show()
```
